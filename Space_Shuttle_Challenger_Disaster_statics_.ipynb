```python
from sklearn.decomposition import PCA
import numpy as np

# Example data: student grades in 3 subjects
data = np.array([[90, 80, 85], [60, 70, 65], [88, 92, 90], [50, 55, 60]])

# PCA for dimensionality reduction
pca = PCA(n_components=1)
reduced_data = pca.fit_transform(data)

print(reduced_data)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Example data: Launch temperatures
data = {
    "Mission": range(1, 26),
    "Temperature": [66, 70, 69, 68, 67, 72, 73, 70, 57, 63, 70, 78, 67, 53, 67, 75, 70, 81, 76, 79, 75, 76, 58, 31, 29]
}

# Create DataFrame
df = pd.DataFrame(data)

# Descriptive statistics
print(df["Temperature"].describe())

# Visualization: Scatter plot
plt.scatter(df["Mission"], df["Temperature"], color='blue')
plt.axhline(y=29, color='red', linestyle='--', label='Challenger Launch (29°F)')
plt.xlabel("Mission")
plt.ylabel("Temperature (°F)")
plt.title("Space Shuttle Launch Temperatures")
plt.legend()
plt.show()
```
